```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

url = "https://raw.githubusercontent.com/Lok997/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Number of pages:", len(df))
print("Median CTR:", df["ctr"].median())
print("Median average position:", df["avg_position"].median())
print("Median impressions over 90 days:", df["impressions_90d"].median())
```

The starter dataset contains 30,000 pages. The median CTR is 7%, the median average position is 10.8, and the median number of impressions over 90 days is 731. These numbers show that there is enough variation in search visibility and click performance to investigate which observable factors are associated with stronger CTR.
