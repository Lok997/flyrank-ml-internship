## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

Data source:

This capstone uses the full pseudonymized FlyRank internship warehouse release hosted on Hugging Face.

Tables used:

- dim_clients — client-level metadata and available history.
- dim_content — content-level metadata and content attributes.
- fact_content_daily_performance — daily search-performance observations.
- fact_content_query_90d — query-level performance signals aggregated over 90 days.

Time window:

The analysis uses historical search-performance data available in the release. Feature windows are defined before the outcome window where prediction is used.

What was excluded:

Client names, domains, URLs, private queries, credentials, and raw exports are excluded from the public analysis. Future or outcome-derived fields are also excluded from predictive features to reduce target leakage.

Data handling:

DuckDB is used to query the hosted Parquet data, and only aggregated feature tables needed for the analysis are brought into pandas. The full warehouse is not downloaded into memory.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

Methodology:

Approach:

The analysis uses observed search-performance signals to rank content items for review and improvement.

Features:

The main signals include recent impressions, previous-period impressions, clicks, sessions, search visibility, query-level signals, content freshness, and ranking position.

Label definition:

For predictive analysis, a decline label is defined using a future performance window. Features are calculated only from information available before that outcome window.

Baseline:

The Week-4 rule-based baseline prioritizes content with meaningful recent impression decline while requiring enough historical impression volume to make the recommendation actionable.

Validation:

Model performance is compared with the baseline on the same held-out data. Where client-level generalization is tested, client_hash_id is used as the grouping variable so that clients do not unnecessarily appear in both training and test groups.

Leakage checks:

Future-window outcome fields and other outcome-derived variables are excluded from the feature set. The analysis does not use client names, domains, URLs, credentials, private queries, or raw exports.
